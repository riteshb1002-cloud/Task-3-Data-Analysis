```python
import pandas as pd
import matplotlib.pyplot as plt

sales_data = {
    "Date": [
        "2026-01-01", "2026-01-02", "2026-01-03",
        "2026-01-04", "2026-01-05", "2026-01-06",
        "2026-01-07", "2026-01-08"
    ],
    "Product": [
        "Laptop", "Mobile", "Headphones", "Laptop",
        "Mobile", "Keyboard", "Headphones", "Mobile"
    ],
    "Category": [
        "Electronics", "Electronics", "Accessories",
        "Electronics", "Electronics", "Accessories",
        "Accessories", "Electronics"
    ],
    "Quantity": [2, 5, 10, 1, 3, 8, 6, 4],
    "Unit_Price": [50000, 15000, 2000, 50000,
                   15000, 1000, 2000, 15000]
}

df = pd.DataFrame(sales_data)
df["Revenue"] = df["Quantity"] * df["Unit_Price"]

df.to_csv("sales_data.csv", index=False)

print("Dataset created successfully!")
df.head()
```

```python
df = pd.read_csv("sales_data.csv")

print("Dataset loaded successfully")
display(df)
```

```python
# Calculate total revenue
total_revenue = df["Revenue"].sum()

# Calculate total quantity sold
total_quantity = df["Quantity"].sum()

# Calculate revenue by product
product_sales = (
    df.groupby("Product")["Revenue"]
      .sum()
      .sort_values(ascending=False)
)

print("Total Revenue:", total_revenue)
print("Total Quantity Sold:", total_quantity)

print("\nRevenue by Product:")
display(product_sales)
```

```python
plt.figure(figsize=(9, 5))

product_sales.plot(kind="bar")

plt.title("Total Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("revenue_by_product.png", dpi=300)
plt.show()
```

```python
quantity_sales = (
    df.groupby("Product")["Quantity"]
      .sum()
      .sort_values(ascending=False)
)

quantity_sales.plot(
    kind="bar",
    figsize=(9, 5),
    title="Quantity Sold by Product"
)

plt.xlabel("Product")
plt.ylabel("Quantity Sold")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("quantity_by_product.png", dpi=300)
plt.show()
```

```python
category_sales = (
    df.groupby("Category")["Revenue"]
      .sum()
      .sort_values(ascending=False)
)

category_sales.plot(
    kind="bar",
    figsize=(8, 5),
    title="Revenue by Category"
)

plt.xlabel("Category")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("revenue_by_category.png", dpi=300)
plt.show()
```
